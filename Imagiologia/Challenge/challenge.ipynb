```python
# Exemplo de carregamento dos datasets reais
df_tumor_mask_real = pd.read_csv("dataset_syntatic_class/CSV1_UC_tumorMask_clean.csv")
df_brain_mask_real = pd.read_csv("dataset_syntatic_class/CSV1_UC_brainMask_clean.csv")
df_whole_scan_mask_real = pd.read_csv("dataset_syntatic_class/CSV1_UC_WholeScanMask_clean.csv")
# Repita esse processo para os demais arquivos CSV
```

```python
# Exemplo de integração dos datasets reais com os sintéticos
# Supondo que você tenha uma coluna "classe_alvo" que indica se o exame é real ou sintético nos datasets sintéticos
df_tumor_mask_real["classe_alvo"] = 1  # Atribua 1 para indicar que os dados são reais
df_tumor_mask_synthetic["classe_alvo"] = 0  # Atribua 0 para indicar que os dados são sintéticos

# Concatene os DataFrames dos datasets reais e sintéticos
df_combined = pd.concat([df_tumor_mask_real, df_tumor_mask_synthetic], ignore_index=True)
```

```python
from sklearn.model_selection import train_test_split


# Divisão dos dados em conjunto de treinamento e teste
X_combined = df_combined.drop(columns=["Volume"])
y_combined = df_combined["classe_alvo"]
X_train_combined, X_test_combined, y_train_combined, y_test_combined = train_test_split(X_combined, y_combined, test_size=0.2, random_state=42)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import GridSearchCV

# Inicialização e treinamento do modelo
model_combined = RandomForestClassifier()
# Definição do espaço de busca dos hiperparâmetros
param_grid = {
    'n_estimators': [50, 100, 200],  # Testa diferentes números de estimadores
    'max_depth': [None, 10, 20],       # Testa diferentes profundidades máximas da árvore
    'criterion': ['gini', 'entropy']  # Testa diferentes critérios de divisão
}
# Inicialização do GridSearchCV para encontrar os melhores hiperparâmetros
grid_search = GridSearchCV(estimator=model_combined, param_grid=param_grid, cv=5, scoring='accuracy')

# # Treinamento do modelo usando busca em grade
grid_search.fit(X_train_combined, y_train_combined)

# Obter o melhor modelo encontrado pelo GridSearchCV
best_model = grid_search.best_estimator_
from sklearn.metrics import accuracy_score

# Avaliação do modelo
y_pred_combined = best_model.predict(X_test_combined)
accuracy_combined = accuracy_score(y_test_combined, y_pred_combined)
print("Acurácia do modelo com dados combinados:", accuracy_combined)
```
