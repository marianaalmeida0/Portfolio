## Advanced examples

```python
f = open("/content/cal112", "r", encoding="UTF-8")
texto = f.read()
slot_info = texto.split(",")
print(slot_info)
name = slot_info[0].strip()
sex = slot_info[1].lower().strip()
age = slot_info[2].strip()
consciousness = slot_info[3].strip()
location = "".join(slot_info[4:]).strip()
print(location)

if sex == "f":
  sex= "feminino"

elif sex=="m":
  sex="masculino"

if "sim" in consciousness:
  consciousness= "está consciente"~

elif "nao" in consciousness:
  consciousness= "não está consciente"

text_prompt = f'''
        Olá, eu sou Mia, um chatbot de suporte a acidentes automóveis. Ocorreu um acidente automóvel com as seguintes informações: o nome da vítima é {name},
         sexo {sex}, idade {age}, estado {consciousness}, no local {location}. Obrigada
    '''
audio_array = generate_audio(text_prompt)
Audio(audio_array, rate=SAMPLE_RATE)
```
